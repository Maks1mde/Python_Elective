```python
source_df = pd.read_csv("titanic.csv")

male_df = source_df[source_df['Sex'] == 'male'].copy()
female_df = source_df[source_df['Sex'] == 'female'].copy()

for part in (male_df, female_df):
    part['Age'] = part['Age'].fillna(part['Age'].median())
    part['Cabin'] = part['Cabin'].fillna('Unknown')
    part['Embarked'] = part['Embarked'].fillna('Unknown')

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
male_df['Age'].hist(bins=20, color='steelblue', edgecolor='black')
plt.title('Age distribution for male')
plt.xlabel('Age')
plt.ylabel('Count')

plt.subplot(1, 2, 2)
female_df['Age'].hist(bins=20, color='salmon', edgecolor='black')
plt.title('Age distribution for female')
plt.xlabel('Age')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

stats = pd.DataFrame({
    'male': {
        'min': male_df['Age'].min(),
        'max': male_df['Age'].max(),
        'mean': male_df['Age'].mean(),
        'mode': male_df['Age'].mode().iloc[0],
        'median': male_df['Age'].median(),
    },
    'female': {
        'min': female_df['Age'].min(),
        'max': female_df['Age'].max(),
        'mean': female_df['Age'].mean(),
        'mode': female_df['Age'].mode().iloc[0],
        'median': female_df['Age'].median(),
    }
})
print('Statistics:')
print(stats)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

male_df['Age'].plot(kind='line', ax=axes[0, 0], title='Line plot for male age', color='royalblue')
axes[0, 0].set_xlabel('Row index')
axes[0, 0].set_ylabel('Age')

female_df['Age'].plot(kind='hist', bins=20, ax=axes[0, 1], title='Histogram for female age', color='tomato')
axes[0, 1].set_xlabel('Age')

male_df['Age'].plot(kind='box', ax=axes[1, 0], title='Box plot for male age')

female_df['Age'].plot(kind='density', ax=axes[1, 1], title='Density for female age', color='darkred')
axes[1, 1].set_xlabel('Age')

plt.tight_layout()
plt.show()

```
